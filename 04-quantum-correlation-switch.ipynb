```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

simulator = AerSimulator()

# Part A: Correlated outcomes
qc_A = QuantumCircuit(2, 2)

qc_A.h(0)
qc_A.cx(0, 1)
qc_A.measure([0, 1], [0, 1])

display(qc_A.draw("mpl"))

print("Prediction: outcomes 00 and 11")
```

```python
job_A = simulator.run(
    transpile(qc_A, simulator),
    shots=100
)

counts_A = job_A.result().get_counts()

print("Part A results:", counts_A)

plot_histogram(counts_A)
plt.show()
```

```python
# Part B: Opposite outcomes
qc_B = QuantumCircuit(2, 2)

qc_B.h(0)
qc_B.cx(0, 1)
qc_B.x(0)
qc_B.measure([0, 1], [0, 1])

display(qc_B.draw("mpl"))

print("Prediction: outcomes 01 and 10")
```

```python
job_B = simulator.run(
    transpile(qc_B, simulator),
    shots=100
)

counts_B = job_B.result().get_counts()

print("Part B results:", counts_B)

plot_histogram(counts_B)
plt.show()
```

```python
print("Part A - Correlated outcomes:", counts_A)
print("Part B - Opposite outcomes:", counts_B)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

plot_histogram(counts_A, ax=axes[0])
axes[0].set_title("Part A: Correlated")

plot_histogram(counts_B, ax=axes[1])
axes[1].set_title("Part B: Opposite")

plt.tight_layout()
plt.show()
```
